## Process stadestér data for the first interval

### Subtask:
Extract the population observations for the time interval -3000 to 1800 from the Stadestér data. Create a DataFrame for plotting this interval.


**Reasoning**:
Transpose the Stadestér DataFrame, filter for non-null coordinates, and iterate through the filtered data to extract population observations within the -3000 to 1800 time interval. Store the city name, coordinates, and observation count in a list. Finally, create a DataFrame from this list and display its head.



```python
df_stadester_transposed = df_stadester.T
df_stadester_filtered_coords = df_stadester_transposed[df_stadester_transposed['coords'].notnull()].copy()

stadester_observation_counts_first_interval = []

for index, row in df_stadester_filtered_coords.iterrows():
    name = row['name']
    coords = row['coords']
    latitude = coords[0]
    longitude = coords[1]
    population_data = row['population']

    observation_count = 0
    if isinstance(population_data, dict):
        for year_str in population_data.keys():
            try:
                year = int(year_str)
                if -3000 <= year <= 1800:
                    observation_count += 1
            except ValueError:
                continue

    if observation_count > 0:
        stadester_observation_counts_first_interval.append({
            'name': name,
            'latitude': latitude,
            'longitude': longitude,
            'observation_count': observation_count
        })

df_stadester_plot_first_interval = pd.DataFrame(stadester_observation_counts_first_interval)
display(df_stadester_plot_first_interval.head())
```
